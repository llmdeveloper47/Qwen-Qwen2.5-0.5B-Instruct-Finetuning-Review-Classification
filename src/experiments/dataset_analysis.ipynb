# Dataset 2 : Amazon 5 x 5

```python
amazon_five_path = '/Users/mle/Documents/MLEngineering/SLM_COURSES/amazon_dataset/dataset/amazon5x5'

train_data_path = os.path.join(amazon_five_path, 'amazon_train.jsonl')
test_data_path = os.path.join(amazon_five_path, 'amazon_test.jsonl')    

train_data = pd.read_json(train_data_path, lines=True)
test_data = pd.read_json(test_data_path, lines=True)

dataset = pd.concat([train_data, test_data])
dataset = dataset.drop_duplicates(subset=['text'])
dataset['sub_categories'] = dataset['labels'].apply(lambda x: x[1])

categories = ['arts-crafts-and-sewing_beading-jewelry-making' , 'arts-crafts-and-sewing_knitting-crochet']
dataset = dataset[~dataset['sub_categories'].isin(categories)]
```

```python
dataset['sub_categories'].value_counts()
```

```python
dataset['sub_categories'].nunique()
```

```python
import random
random.seed(42)
train_df = dataset.groupby('sub_categories').sample(n=14000)
temp_df = dataset[~dataset['text'].isin(list(train_df['text']))]
val_df = temp_df.groupby('sub_categories').sample(n=1500)
test_df = temp_df[~temp_df['text'].isin(list(val_df['text']))]
test_df = test_df.groupby('sub_categories').sample(n=450)

print(f'train_df : {train_df.shape}')
print(f'val_df : {val_df.shape}')
print(f'test_df : {test_df.shape}')

train_df.to_csv('/Users/mle/Documents/MLEngineering/SLM_COURSES/amazon_dataset/dataset/final_sampled_data/train_df.csv', index=False)
val_df.to_csv('/Users/mle/Documents/MLEngineering/SLM_COURSES/amazon_dataset/dataset/final_sampled_data/val_df.csv', index=False)
test_df.to_csv('/Users/mle/Documents/MLEngineering/SLM_COURSES/amazon_dataset/dataset/final_sampled_data/test_df.csv', index=False)
```
